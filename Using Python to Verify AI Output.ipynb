# Using Python to Verify AI Output - Avocados 25 & 26

```python
# Import Packages
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
# Create timestamp and author
from datetime import datetime
time_now=datetime.now().strftime('%m_%d_%Y_%I%M%p')
print("Current Time:{}".format(time_now))
print("Adeline Huang")
```

```python
# Loading datasets
df_2025 = pd.read_csv('2025 Data.csv')
df_2026 = pd.read_csv('2026 Data.csv')
```

```python
# Help calculate percentage change
def calc_perc_change(old_val, new_val):
    return ((new_val - old_val) / old_val) * 100
```

```python
# National Pricing and Volume Shifts 
# Filter for Total U.S. geography
us_25 = df_2025[df_2025['Geography'] == 'Total U.S.']
us_26 = df_2026[df_2026['Geography'] == 'Total U.S.']
```

```python
# calculating averages for 2025
us_25_stats = us_25.groupby('Type')[['ASP Current Year', 'Total Bulk and Bags']].mean()
conv_asp_25 = us_25_stats.loc['Conventional', 'ASP Current Year']
conv_vol_25 = us_25_stats.loc['Conventional', 'Total Bulk and Bags']
org_asp_25 = us_25_stats.loc['Organic', 'ASP Current Year']
org_vol_25 = us_25_stats.loc['Organic', 'Total Bulk and Bags']
```

```python
# calculating averages for 2026
us_26_stats = us_26.groupby('Type')[['ASP Current Year', 'Total Bulk and Bags']].mean()
conv_asp_26 = us_26_stats.loc['Conventional', 'ASP Current Year']
conv_vol_26 = us_26_stats.loc['Conventional', 'Total Bulk and Bags']
org_asp_26 = us_26_stats.loc['Organic', 'ASP Current Year']
org_vol_26 = us_26_stats.loc['Organic', 'Total Bulk and Bags']
```

```python
# regional variations, filtering for southeast and great lakes
se_25 = df_2025[df_2025['Geography'] == 'Southeast']
se_26 = df_2026[df_2026['Geography'] == 'Southeast']
gl_25 = df_2025[df_2025['Geography'] == 'Great Lakes']
gl_26 = df_2026[df_2026['Geography'] == 'Great Lakes']
```

```python
# southeast conventional stats
se_conv_asp_25 = se_25[se_25['Type'] == 'Conventional']['ASP Current Year'].mean()
se_conv_asp_26 = se_26[se_26['Type'] == 'Conventional']['ASP Current Year'].mean()
se_conv_vol_25 = se_25[se_25['Type'] == 'Conventional']['Total Bulk and Bags'].mean()
se_conv_vol_26 = se_26[se_26['Type'] == 'Conventional']['Total Bulk and Bags'].mean()
```

```python
# great lakes organic stats
gl_org_vol_25 = gl_25[gl_25['Type'] == 'Organic']['Total Bulk and Bags'].mean()
gl_org_vol_26 = gl_26[gl_26['Type'] == 'Organic']['Total Bulk and Bags'].mean()
```

```python
# to print calculations
print("--- NATIONAL METRICS (Total U.S.) ---")
print(f"Conventional ASP 2025: ${conv_asp_25:.2f} | 2026: ${conv_asp_26:.2f}")
print(f"Conventional ASP % Change: {calc_perc_change(conv_asp_25, conv_asp_26):.1f}%\n")

print(f"Conventional Vol 2025: {conv_vol_25:,.0f} | 2026: {conv_vol_26:,.0f}")
print(f"Conventional Vol % Change: {calc_perc_change(conv_vol_25, conv_vol_26):.1f}%\n")

print(f"Organic ASP 2025: ${org_asp_25:.2f} | 2026: ${org_asp_26:.2f}")
print(f"Organic ASP % Change: {calc_perc_change(org_asp_25, org_asp_26):.1f}%\n")

print(f"Organic Vol 2025: {org_vol_25:,.0f} | 2026: {org_vol_26:,.0f}")
print(f"Organic Vol % Change: {calc_perc_change(org_vol_25, org_vol_26):.1f}%\n")

print("--- REGIONAL METRICS ---")
print(f"Southeast Conventional ASP % Change: {calc_perc_change(se_conv_asp_25, se_conv_asp_26):.1f}%")
print(f"Southeast Conventional Vol % Change: {calc_perc_change(se_conv_vol_25, se_conv_vol_26):.1f}%")
print(f"Great Lakes Organic Vol % Change: {calc_perc_change(gl_org_vol_25, gl_org_vol_26):.1f}%")
```
